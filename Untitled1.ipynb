```python
# ============================================================================
#  shares_outstanding validation -- old resolver vs fixed resolver, live SEC data
#
#  RUN IN GOOGLE COLAB (works in a phone browser):
#    1. open https://colab.research.google.com  ->  "New notebook"
#    2. paste this whole script into the cell
#    3. put your name/email in IDENTITY below (the SEC requires it)
#    4. tap the play button. Expect 10-20 minutes for the full list.
# ============================================================================
IDENTITY = "Your Name you@example.com"      # <-- change this
FILINGS_PER_COMPANY = 10                    # most recent 10-K/10-Q filings each
SELFTEST_ONLY = False                       # True = run the offline replay only

import subprocess, sys
if not SELFTEST_ONLY:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "edgartools==5.28.3"], check=False)

SYMBOLS = [
    "CPK", "AEHR", "ALTO", "AMTX", "ASII", "BDCO", "BEEM", "BMPA", "CBZ", "CCL",
    "CCRN", "CLBK", "DAIO", "DCTH", "AAP", "AAM", "A", "AAME", "CDIX", "CLOW",
    1777946, 1787518,
    "GOOGL", 1067983, "KO", "V", "META", "LEN", "XOM", "UHAL", "F", "RDDT",
    "MKC", "CWEN", "UAA", "NWSA", "FOX", "ZG", "LBRDA", "EXPE",
    "AAPL", "MSFT", "NVDA", "AMZN", "JPM", "WMT", "COST", "JNJ", "PG", "HD",
    "CAT", "DE", "MMM", "T", "VZ", "PFE", "MRK", "ABT", "ORCL", "CSCO",
]

class Concepts:
    ROLES = {
        "shares_outstanding_cover_page": ["dei:EntityCommonStockSharesOutstanding"],
        "shares_outstanding_balance_sheet": ["us-gaap:CommonStockSharesOutstanding",
                                             "us-gaap:SharesOutstanding"],
        "weighted_average_basic": ["us-gaap:WeightedAverageNumberOfSharesOutstandingBasic",
                                   "us-gaap:WeightedAverageNumberOfSharesIssuedBasic"],
        "weighted_average_diluted": ["us-gaap:WeightedAverageNumberOfDilutedSharesOutstanding",
                                     "us-gaap:WeightedAverageNumberOfShareOutstandingBasicAndDiluted"],
        "net_income_to_common": ["us-gaap:NetIncomeLossAvailableToCommonStockholdersBasic",
                                 "us-gaap:NetIncomeLoss", "us-gaap:ProfitLoss"],
        "eps_basic": ["us-gaap:EarningsPerShareBasic", "us-gaap:EarningsPerShareBasicAndDiluted"],
    }
    def role(self, name):
        return self.ROLES.get(name, [])

CONCEPTS = Concepts()

# ===== FIXED resolver (branch claude/xbrl-facts-extraction, commit c5e9e5c) =====
import math
from collections import namedtuple
from datetime import date
MIN_RATIO = 0.2
MAX_RATIO = 20.0
SCALE_ERROR_RATIO = 50.0
MIN_EPS_FOR_REFERENCE = 0.01
MIN_DILUTED_RATIO = 0.25
MAX_DILUTED_RATIO = 10.0
DILUTED_VS_BASIC_FLOOR = 0.99
Resolution = namedtuple('Resolution', 'value rung anchor ratio detail')

def _num(v):
    if v is None:
        return None
    try:
        f = float(v)
    except (TypeError, ValueError):
        return None
    return None if f != f else f

def _str(v):
    if v is None:
        return None
    s = str(v).strip()
    if not s or s.lower() in ('nan', 'none', 'nat'):
        return None
    return s

def _days(rec):
    s = _str(rec.get('start')) or _str(rec.get('period_start'))
    e = _str(rec.get('end')) or _str(rec.get('period_end'))
    if not s or not e:
        return None
    try:
        return (date.fromisoformat(e) - date.fromisoformat(s)).days
    except ValueError:
        return None

def normalize(records):
    seen, out = (set(), [])
    for r in records or []:
        concept = _str(r.get('concept'))
        if not concept:
            continue
        rec = {'concept': concept, 'label': _str(r.get('label')), 'value': _num(r.get('numeric_value')) if r.get('numeric_value') is not None else _num(r.get('value')), 'instant': _str(r.get('period_instant')), 'start': _str(r.get('period_start')), 'end': _str(r.get('period_end')), 'dimensioned': bool(r.get('is_dimensioned')), 'fiscal_period': _str(r.get('fiscal_period'))}
        key = (rec['concept'], rec['label'], rec['value'], rec['instant'], rec['start'], rec['end'], rec['dimensioned'])
        if key in seen:
            continue
        seen.add(key)
        out.append(rec)
    return out

def _pick(records, concept, dimensioned):
    return [r for r in records if r['concept'] == concept and r['dimensioned'] is dimensioned and (r['value'] is not None)]

def _basic_anchor(records, period_of_report, concepts):
    rows = []
    for concept in concepts.role('weighted_average_basic'):
        rows.extend(_pick(records, concept, dimensioned=False))
    if not rows:
        return None
    exact = [r for r in rows if r['end'] == period_of_report]
    pool = exact or rows
    pool = sorted(pool, key=lambda r: (r['end'] or '', -(_days(r) or 10 ** 6)), reverse=True)
    return pool[0]['value']

def _cover_total(recs, concepts):
    cover = []
    for concept in concepts.role('shares_outstanding_cover_page'):
        cover.extend(_pick(recs, concept, dimensioned=False))
    if not cover:
        return None
    cover = sorted(cover, key=lambda r: r['instant'] or '')
    return cover[-1]['value']

def _ratio_of(v, ref):
    if v is None or not ref:
        return None
    return v / ref

def _scale_sized(v, ref):
    r = _ratio_of(v, ref)
    return r is not None and (r >= SCALE_ERROR_RATIO or r <= 1 / SCALE_ERROR_RATIO)

def _by_period(recs, role, concepts):
    out = {}
    for concept in concepts.role(role):
        for r in _pick(recs, concept, dimensioned=False):
            if r['start'] and r['end']:
                out.setdefault((r['start'], r['end']), r['value'])
    return out

def eps_implied_count(recs, period_of_report, concepts):
    ni = _by_period(recs, 'net_income_to_common', concepts)
    eps = _by_period(recs, 'eps_basic', concepts)
    pairs = [(k, ni[k], eps[k]) for k in ni if k in eps and ni[k] and eps[k] and (abs(eps[k]) >= MIN_EPS_FOR_REFERENCE)]
    if not pairs:
        return None
    exact = [p for p in pairs if p[0][1] == period_of_report]
    pool = exact or pairs
    pool.sort(key=lambda p: _days({'start': p[0][0], 'end': p[0][1]}) or 10 ** 6)
    _, n, e = pool[0]
    return abs(n / e)

def resolve_shares_outstanding(records, period_of_report, concepts):
    recs = normalize(records)
    basic = _basic_anchor(recs, period_of_report, concepts)
    implied = eps_implied_count(recs, period_of_report, concepts)
    cover = _cover_total(recs, concepts)
    if implied:
        reference, ref_name = (implied, 'net income / basic EPS')
    elif basic:
        reference, ref_name = (basic, 'weighted-average basic')
    else:
        reference, ref_name = (None, None)
    detail = {'reference': ref_name, 'anchor': reference, 'weighted_average_basic': basic, 'eps_implied': implied, 'cover': cover}
    if implied and basic and _scale_sized(implied, basic):
        detail['basic_conflict'] = f'weighted-average basic {basic:,.0f} is {_ratio_of(basic, implied):.3g}x the count implied by net income / EPS {implied:,.0f}; the average is in the wrong units and is not used'
    rejected = []

    def _refuse(where, v):
        rejected.append({'rung': where, 'value': v, 'reason': f'{_ratio_of(v, reference):.3g}x the {ref_name} {reference:,.0f}'})

    def _in_band(v):
        r = _ratio_of(v, reference)
        return r is None or MIN_RATIO <= r <= MAX_RATIO
    bs_rows = []
    for concept in concepts.role('shares_outstanding_balance_sheet'):
        bs_rows.extend(_pick(recs, concept, dimensioned=False))
    r1 = [r for r in bs_rows if r['instant'] == period_of_report]
    detail['balance_sheet_candidates'] = [r['value'] for r in bs_rows]
    outvote = not implied and basic and cover and (cover > 0)
    survivors, seen = ([], set())
    for r in r1:
        v = r['value']
        if v in seen:
            continue
        seen.add(v)
        if v <= 0:
            rejected.append({'rung': 1, 'value': v, 'reason': 'not a positive count'})
        elif _in_band(v):
            survivors.append(v)
        elif outvote and 0.5 <= v / cover <= 2 and _scale_sized(v, basic):
            detail['basic_outvoted'] = f'balance-sheet line {v:,.0f} and cover count {cover:,.0f} agree; weighted-average basic {basic:,.0f} is {_ratio_of(basic, v):.3g}x both and is treated as the one in the wrong units'
            survivors.append(v)
        else:
            _refuse(1, v)
    if survivors:
        if len(survivors) > 1 and reference:
            survivors.sort(key=lambda v: abs(math.log(v / reference)))
        if len(survivors) > 1:
            detail['balance_sheet_alternatives'] = survivors[1:]
        detail['rejected'] = rejected
        v = survivors[0]
        return Resolution(v, 1, reference, _ratio_of(v, reference), detail)
    if cover is not None and cover > 0:
        if _in_band(cover):
            detail['rejected'] = rejected
            return Resolution(cover, 2, reference, _ratio_of(cover, reference), detail)
        _refuse(2, cover)
    dim = []
    for concept in concepts.role('shares_outstanding_cover_page'):
        dim.extend(_pick(recs, concept, dimensioned=True))
    detail['rejected'] = rejected
    if dim:
        latest = max((r['instant'] or '' for r in dim))
        at_latest = [r for r in dim if (r['instant'] or '') == latest]
        total = sum((r['value'] for r in at_latest))
        detail['class_values'] = [r['value'] for r in at_latest]
        detail['class_instant'] = latest
        if not reference:
            rejected.append({'rung': 3, 'value': total, 'reason': 'no reference count to anchor against; classes are not demonstrably equivalent'})
        elif not _in_band(total):
            _refuse(3, total)
            rejected[-1]['reason'] += '; classes are not economically equivalent'
        else:
            return Resolution(total, 3, reference, total / reference, detail)
    if not bs_rows and cover is None and (not dim):
        detail['reason'] = 'no share-count facts'
    elif not rejected:
        detail['reason'] = 'no candidate at the period end and no cover-page facts'
    else:
        detail['reason'] = 'every candidate refused; see rejected'
    return Resolution(None, None, reference, None, detail)

def resolve_diluted_average(records, period_of_report, concepts, fiscal_period=None, cross_check=None):
    recs = normalize(records)
    rows = []
    for concept in concepts.role('weighted_average_diluted'):
        rows.extend(_pick(recs, concept, dimensioned=False))
    if not rows:
        return Resolution(None, None, None, None, {'reason': 'no undimensioned diluted fact'})
    ending = [r for r in rows if r['end'] == period_of_report] or rows
    if fiscal_period:
        matched = [r for r in ending if r['fiscal_period'] == fiscal_period]
        ending = matched or ending
    ending = sorted(ending, key=lambda r: _days(r) if _days(r) is not None else 10 ** 6)
    value = ending[0]['value']
    detail = {'days': _days(ending[0]), 'end': ending[0]['end'], 'candidate': value}
    basic = _basic_anchor(recs, period_of_report, concepts)
    if basic:
        detail['basic'] = basic
        detail['vs_basic'] = value / basic
        if value < basic * DILUTED_VS_BASIC_FLOOR:
            detail['reason'] = f'diluted {value:,.0f} is below basic {basic:,.0f}; either a partial figure or a period mismatch'
            return Resolution(None, None, basic, value / basic, detail)
    if cross_check:
        ratio = value / cross_check
        detail['ratio'] = ratio
        if not MIN_DILUTED_RATIO <= ratio <= MAX_DILUTED_RATIO:
            detail['reason'] = f'diluted {value:,.0f} is {ratio:.2f}x shares outstanding {cross_check:,.0f}; outside the plausible range'
            return Resolution(None, None, cross_check, ratio, detail)
        return Resolution(value, 'diluted', cross_check, ratio, detail)
    return Resolution(value, 'diluted', None, None, detail)

# ===== OLD resolver: production's decision rule, the code path that wrote your parquet =====
def old_resolve_shares_outstanding(records, period_of_report, concepts):
    recs = normalize(records)
    anchor = _basic_anchor(recs, period_of_report, concepts)
    bs_rows = []
    for concept in concepts.role("shares_outstanding_balance_sheet"):
        bs_rows.extend(_pick(recs, concept, dimensioned=False))
    r1 = [r for r in bs_rows if r["instant"] == period_of_report]
    if r1:
        v = r1[0]["value"]
        return Resolution(v, 1, anchor, (v / anchor) if anchor else None, {})
    cover = []
    for concept in concepts.role("shares_outstanding_cover_page"):
        cover.extend(_pick(recs, concept, dimensioned=False))
    cover = sorted(cover, key=lambda r: r["instant"] or "")
    if cover:
        v = cover[-1]["value"]
        return Resolution(v, 2, anchor, (v / anchor) if anchor else None, {})
    dim = []
    for concept in concepts.role("shares_outstanding_cover_page"):
        dim.extend(_pick(recs, concept, dimensioned=True))
    if not dim or not anchor:
        return Resolution(None, None, anchor, None, {})
    latest = max((r["instant"] or "") for r in dim)
    total = sum(r["value"] for r in dim if (r["instant"] or "") == latest)
    ratio = total / anchor
    if not (MIN_RATIO <= ratio <= MAX_RATIO):
        return Resolution(None, None, anchor, ratio, {})
    return Resolution(total, 3, anchor, ratio, {})

# ===== offline replay of CPK as your live run showed it =====
def selftest():
    por, cov = "2023-09-30", "2023-11-01"
    def inst(c, v, i=por, dim=False): return {"concept": c, "numeric_value": v, "period_instant": i, "is_dimensioned": dim}
    def dur(c, v, s="2023-07-01", e=por): return {"concept": c, "numeric_value": v, "period_start": s, "period_end": e, "is_dimensioned": False}
    cpk = [inst("us-gaap:CommonStockSharesOutstanding", 17_796_741_000),
           inst("dei:EntityCommonStockSharesOutstanding", 17_796_741_000, cov),
           dur("us-gaap:WeightedAverageNumberOfSharesOutstandingBasic", 17_790_000),
           dur("us-gaap:WeightedAverageNumberOfDilutedSharesOutstanding", 17_852_024),
           dur("us-gaap:NetIncomeLoss", 12_500_000.0), dur("us-gaap:EarningsPerShareBasic", 0.70)]
    o = old_resolve_shares_outstanding(cpk, por, CONCEPTS)
    n = resolve_shares_outstanding(cpk, por, CONCEPTS)
    od = resolve_diluted_average(cpk, por, CONCEPTS, cross_check=o.value)
    nd = resolve_diluted_average(cpk, por, CONCEPTS, cross_check=n.value)
    assert o.value == 17_796_741_000 and od.value is None, "old replica does not reproduce the parquet"
    assert n.value is None and nd.value == 17_852_024, "fix does not resolve CPK"
    assert n.detail["reference"] == "net income / basic EPS"
    print("selftest OK: old -> 17,796,741,000 with diluted null (as in the parquet); "
          "fixed -> both x1000 instants refused against net income / EPS, shares null, diluted 17,852,024 kept")

selftest()
if SELFTEST_ONLY:
    raise SystemExit(0)

# ===== live run =====
import time
from edgar import set_identity, Company
set_identity(IDENTITY)

def facts_of(xbrl):
    out = []
    for f in xbrl.facts.get_facts():
        c = f.get("concept")
        if not c:
            continue
        out.append({"concept": str(c).replace("_", ":", 1) if ":" not in str(c) else str(c),
                    "numeric_value": f.get("numeric_value"), "value": f.get("value"),
                    "period_instant": f.get("period_instant"), "period_start": f.get("period_start"),
                    "period_end": f.get("period_end"), "is_dimensioned": f.get("is_dimensioned"),
                    "fiscal_period": f.get("fiscal_period"), "label": f.get("label")})
    return out

def fmt(v):
    return "-" if v is None else f"{v:,.0f}"

def rat(v, a):
    return "-" if (v is None or not a) else f"{v / a:.3g}x"

def steps(series):
    vals = [v for v in series if v]
    return sum(1 for a, b in zip(vals, vals[1:]) if a / b >= 50 or b / a >= 50)

rows_all, t0 = [], time.time()
for sym in SYMBOLS:
    try:
        co = Company(sym)
        filings = co.get_filings(form=["10-K", "10-Q"]).head(FILINGS_PER_COMPANY)
    except Exception as e:
        print(f"\n== {sym}: could not load company ({type(e).__name__}: {e})")
        continue
    name = getattr(co, "name", sym)
    print(f"\n== {sym}  {name}")
    print(f"   {'period':10} {'form':6} {'OLD shares':>18} {'old/ref':>8} {'FIXED shares':>18} {'new/ref':>8} rung ref  {'OLD diluted':>16} {'FIXED diluted':>16}")
    rows = []
    for f in filings:
        try:
            x = f.xbrl()
            if x is None:
                continue
            por = str(getattr(x, "period_of_report", "") or "")
            recs = facts_of(x)
            o = old_resolve_shares_outstanding(recs, por, CONCEPTS)
            n = resolve_shares_outstanding(recs, por, CONCEPTS)
            od = resolve_diluted_average(recs, por, CONCEPTS, cross_check=o.value)
            nd = resolve_diluted_average(recs, por, CONCEPTS, cross_check=n.value)
            rows.append((por, f.form, o.value, n.value, n.rung, n.anchor, od.value, nd.value))
            ref = {"net income / basic EPS": "eps", "weighted-average basic": "avg", None: "-"}[n.detail.get("reference")]
            print(f"   {por:10} {f.form:6} {fmt(o.value):>18} {rat(o.value, n.anchor):>8} {fmt(n.value):>18} {rat(n.value, n.anchor):>8} {str(n.rung):>4} {ref:4} {fmt(od.value):>16} {fmt(nd.value):>16}")
        except Exception as e:
            print(f"   {getattr(f, 'accession_no', '?')}: {type(e).__name__}: {e}")
    rows.sort()
    for r in rows:
        rows_all.append((sym,) + r)
    print(f"   power-of-ten steps: OLD {steps([r[2] for r in rows])}  FIXED {steps([r[3] for r in rows])}"
          f"   | diluted null: OLD {sum(r[6] is None for r in rows)}  FIXED {sum(r[7] is None for r in rows)}"
          f"   ({time.time() - t0:.0f}s)")

# ===== summary =====
def off(v, a):
    return bool(v is not None and a and not (0.2 <= v / a <= 20))

n = len(rows_all)
print("\n" + "=" * 78)
print(f"{n} filings across {len({r[0] for r in rows_all})} companies")
print("share counts more than 20x away from the reference (net income / EPS, else weighted-average basic):")
print(f"   OLD   {sum(off(r[3], r[6]) for r in rows_all)}")
print(f"   FIXED {sum(off(r[4], r[6]) for r in rows_all)}")
print("power-of-ten steps between consecutive filings, summed over companies:")
by = {}
for r in rows_all:
    by.setdefault(r[0], []).append(r)
print(f"   OLD   {sum(steps([r[3] for r in v]) for v in by.values())}")
print(f"   FIXED {sum(steps([r[4] for r in v]) for v in by.values())}")
print(f"share count null:        OLD {sum(r[3] is None for r in rows_all)}   FIXED {sum(r[4] is None for r in rows_all)}")
print(f"diluted average null:    OLD {sum(r[7] is None for r in rows_all)}   FIXED {sum(r[8] is None for r in rows_all)}")
changed = [r for r in rows_all if r[3] != r[4]]
print(f"\nfilings where the fixed value differs from the old one: {len(changed)}")
for r in changed:
    factor = (r[4] / r[3]) if (r[3] and r[4]) else None
    print(f"   {r[0]:8} {r[1]} {r[2]:6} old={fmt(r[3]):>18} new={fmt(r[4]):>18}"
          + (f"  ({factor:.4g}x)" if factor else "") + f"  rung {r[5]}")
```
